````python
def odd_numbers(input_list: list):
    a = []
    for i in input_list:
        if type(i) in [int, bool]:
            if i % 2 != 0:
                a.append(i)
        if type(i) in [float, complex]:
            if (i + 1) % 2 == 0:
                a.append(i)
    return a
````

````python
def longest_string(input_list: list):
    length = 0
    for i in input_list:
        if isinstance(i, str):
            if len(i) > length:    
                length = len(i)
    if length > 0:    
        return length
    else:
        return None
````

````python
def unflatten(input_list: list, control_tuple: tuple):
    b = []
    c = 0
    for i in control_tuple:
        a = []
        for j in range(i):
            if len(input_list) > c + j:
                a.append(input_list[c + j])
            else:
                a.append(None)
        b.append(a)
        c += i
    return b
````

````python
def elements_in_common(input_list1: list, input_list2: list):
    a = []
    for i in input_list1:
        for j in input_list2:
            if j is i:
                a.append(j)
    return a
    
````

````python
class Counter:
    def __init__(self):
        self.count = 0
    
    def increment(self):
        self.count += 1
    
    def reset(self):
        self.count = 0
````

````python
class BankAccount:
    def __init__(self, balance):
        if balance < 0:
            raise ValueError("cannot have a negative balance")
        self.balance = balance
    
    def deposit(self, amount):
        if amount < 0:
            raise ValueError("deposits must be positive")
        self.balance += amount
    
    def withdraw(self, amount):
        if amount > self.balance:
            raise ValueError("amount exceeds balance")
        if amount < 0:
            raise ValueError("withdrawals must be positive")
        self.balance -= amount
    
    def __repr__(self):
        return f"BankAccount({self.balance})"
````

````python
a = BankAccount(50)
b = BankAccount(9999)
print(a, b)          # Bank account class Bank account class
````

````python
'''clean claude version (reference post coding)'''

class RestaurantTable:
    """A table in a restaurant, which may be reserved and has a fixed location."""

    VALID_LOCATIONS = frozenset({"inside", "outside"})
    MAX_SPARE_SEATS = 2

    def __init__(self, seats: int, location: str) -> None:
        if type(seats) is not int or seats < 1:
            raise ValueError(f"seats must be a positive integer, got {seats!r}")
        if location not in self.VALID_LOCATIONS:
            raise ValueError(
                f"location must be one of {sorted(self.VALID_LOCATIONS)}, got {location!r}"
            )
        self.seats = seats
        self.location = location
        self.is_reserved = False

    def reserve(self) -> None:
        self.is_reserved = True

    def is_suitable(self, party_size: int, preference: str | None = None) -> bool:
        if type(party_size) is not int or party_size < 1:
            raise ValueError(f"party size must be a positive integer, got {party_size!r}")
        if preference is not None and preference not in self.VALID_LOCATIONS:
            raise ValueError(
                f"preference must be one of {sorted(self.VALID_LOCATIONS)}, got {preference!r}"
            )

        spare_seats = self.seats - party_size
        return (
            not self.is_reserved
            and 0 <= spare_seats <= self.MAX_SPARE_SEATS
            and (preference is None or preference == self.location)
        )

    def __repr__(self) -> str:
        return (
            f"<RestaurantTable seats={self.seats} "
            f"location={self.location!r} reserved={self.is_reserved}>"
        )
````
